```python
# write a program without using functions to generate a table of a given number
number = int (input("Enter a number to generate the table for: "))

for i in range(1,11):
    mult = i * number
    print (f"{i} x {number} = {mult}")
```

```python
# write a program using functions to generate a table of a given number
number = int (input("Enter a number to generate the table for: "))

# function definition
def tbl (n):
    for i in range(1,11):
        mult = i * n
        print (f"{i} x {n} = {mult}")

#function calling
tbl(number)
# function calling for 10
print ("function calling for 10")
tbl(10)
```

```python
# write a program using functions to retun the sum and product of 2 numbers
n1 = int (input("Enter the 1st number: "))
n2 = int (input("Enter the 2nd number: "))

# function definition
def multproduct (x, y):
    sm = x + y
    mlt = x * y
    return sm, mlt

#function calling
s, m = multproduct(n1,n2)
# function calling for 10
print (f"Sum of the numbers = {s}")
print (f"Product of the numbers = {m}")
```

# Case Study: Calculating Final Marks
- <font size=3>This case study focuses on building a function-based program to calculate a student's final mark for the module, which includes two assessments:</font>
    - <font size=2>AE1: 40% of the total grade</font>
    - <font size=2>AE2: 60% of the total grade</font>
- <font size=3>The final mark is calculated by combining the weighted scores of both assessments.</font>
    - <font size=2>For example, if a student scores 42 in AE1 and 35 in AE2, the calculation is:</font>
         - <font size=2>(42 × 0.4) + (35 × 0.6) = 37.8</font>
    - <font size=2>This result is then rounded to the nearest whole number, giving a final mark of 38</font>

## Steps to Develop a Function-Based Program
- <font size=3>Plan the Process</font>
    - <font size=2>Think through the steps needed to calculate the total mark</font>
    - <font size=2>Identify the required inputs: number of assessments, each assessment’s weight, and score.</font>
    - <font size=2>Choose the suitable data type or collection to help the program run efficiently and handle data properly.</font>
- <font size=3>Write a Simple Sequential Version</font>
    - <font size=2>Implement the process in a basic, step-by-step format without using functions.</font>
    - <font size=2>This helps test the logic and understand how the data flows.</font>
- <font size=3>Identify Reusable Tasks</font>
    - <font size=2>Look for parts of the code that perform a specific task (e.g., get assessment detail, calculating total mark).</font>
    - <font size=2>These are good candidates to be turned into functions.</font>
- <font size=3>Define Functions</font>
    - <font size=2>Use def to define functions for each task.</font>
    - <font size=2>Decide what parameters each function needs and what value it should return.</font>
- <font size=3>Refactor the Code</font>
    - <font size=2>Replace the repeated or specific logic in the sequential code with function calls.</font>
    - <font size=2>This makes the program cleaner, easier to read, and reusable.</font>
- <font size=3>Test and Improve</font>
    - <font size=2>Run the program with different inputs to make sure it works correctly.</font>
    - <font size=2>Adjust the functions or structure as needed for clarity or flexibility.</font>

# Simple Sequential Version

```python
# Set intial variables
assessment_details = {}
total_marks = 0

# get the assessments details
num_assessments = int (input ("Enter the numbre of assessments: "))

for i in range(num_assessments):
    assessment_weight = []
    component = input ("Enter an assessment component [e.g. AE1, AE2]:")
    weight = int (input("Enter the assesment weight (e.g. 40, 60): "))
    weight_percentage = weight / 100
    assessment_weight.append(weight_percentage)

    assessment_marks = int (input (f"Enter your assessment marks for {component}:"))
    assessment_weight.append(assessment_marks)
    assessment_details[component.upper()] = assessment_weight

#calculate the total marks
for key, value in assessment_details.items():
    print (f"Assesment {key}:")
    weight = value[0]
    score = value[1]
    mark = weight * score
    print (f"Marks for {key}: {mark}:")
    total_marks = total_marks + mark

# Display the total marks
print (f"The total marks are: {round(total_marks)}")
```

# Function, Parameters, Arguments and Return Values

## Parameters

- <font size ="3"><b>A parameter</b> (also known as <b>a formal argument</b>) is an input to a function</font>
- <font size ="3">Parameters allow us to pass information into functions</font>
- <font size="3">Listed as variables inside parentheses at the end of the function name</font>

## Arguments

- <font size ="3"><b>An argument</b> (also known as <b>an actual argument</b>) is data that is passed to a function</font>
- <font size ="3">Python functions are invoked through call-by-object-reference - values are bound to objects and references to the objects are passed when a function is invoked</font>

## Retrun values
- <font size ="3">You can return data from a function using a <b><span style="color: green;">return</span></b> statement at the end of a function</font>
- <font size ="3">Return value: data received by the main program from a function, which can be used for calculations or other operations</font>

### Function, Parameters, Arguments, and Return Value Code

```python
# Convert the simple code into modular or function based code
assessment_details = {}

# get the assessments details
def Getting_Details():
    num_assessments = int (input ("Enter the numbre of assessments: "))

    for i in range(num_assessments):
        assessment_weight = []
        component = input ("Enter an assessment component [e.g. AE1, AE2]:")
        weight = int (input("Enter the assesment weight (e.g. 40, 60): "))
        weight_percentage = weight / 100
        assessment_weight.append(weight_percentage)

        assessment_marks = int (input (f"Enter your assessment marks for {component}:"))
        assessment_weight.append(assessment_marks)
        assessment_details[component.upper()] = assessment_weight

#calculate the total marks
def Calculate():
    total_marks = 0
    for key, value in assessment_details.items():
        print (f"Assesment {key}:")
        weight = value[0]
        score = value[1]
        mark = weight * score
        print (f"Marks for {key}: {mark}:")
        total_marks = total_marks + mark

    return total_marks

# Display the total marks
def Display(m):
    print (f"The total marks are: {round(m)}")

# Calling the functions
# Calling Getting_Details Function
Getting_Details()

# Calling Calculate Function
marks = Calculate()

# Calling Display Function
Display(marks)
```

# Define the Program’s Starting Point
- <font size=3>To keep your program organised and easy to understand, it's a good practice to clearly separate the main execution part of your code.</font>
- <font size=3>This ensures that the main part of your program only runs when the file is executed directly, not when it's imported as a module in another script. It helps make your code more reusable and easier to test.</font>


<div class="alert alert-block alert-info">
<b>if __name__ == "__main__"</b> <br>
    <ul>
        <li>Syntactically, Python’s if __name__ == "__main__" idiom is just a normal conditional block</li>
        <li>It allows you to execute code when the file runs as a Script, but not when it’s imported as a Module</li>
    </ul>
</div>

```python
# Convert the simple code into modular or function based code
assessment_details = {}

# get the assessments details
def Getting_Details():
    num_assessments = int (input ("Enter the numbre of assessments: "))

    for i in range(num_assessments):
        assessment_weight = []
        component = input ("Enter an assessment component [e.g. AE1, AE2]:")
        weight = int (input("Enter the assesment weight (e.g. 40, 60): "))
        weight_percentage = weight / 100
        assessment_weight.append(weight_percentage)

        assessment_marks = int (input (f"Enter your assessment marks for {component}:"))
        assessment_weight.append(assessment_marks)
        assessment_details[component.upper()] = assessment_weight

#calculate the total marks
def Calculate():
    total_marks = 0
    for key, value in assessment_details.items():
        print (f"Assesment {key}:")
        weight = value[0]
        score = value[1]
        mark = weight * score
        print (f"Marks for {key}: {mark}:")
        total_marks = total_marks + mark

    return total_marks

# Display the total marks
def Display(m):
    print (f"The total marks are: {round(m)}")

# Calling the functions in the main executable area of the code

if __name__ == "__main__":
    # Calling Getting_Details Function
    Getting_Details()

    # Calling Calculate Function
    marks = Calculate()

    # Calling Display Function
    Display(marks)
```

# Functions and File Handling
- <font size=3>Case Study: Calculating Final Marks from a CSV File</font>
- <font size=3>We’re working with a dataset for the COM731 module, which includes two assessments:</font>
    - <font size=2>AE1 (40% weight)</font>
    - <font size=2>AE2 (60% weight)</font>
- <font size=3>The goal is to:</font>
    - <font size=2>Read student data from a CSV file.</font>
    - <font size=2>Calculate each student’s final mark using the assessment weights.</font>
    - <font size=2>Round the final mark.</font>
    - <font size=2>Determine if the student passes (mark ≥ 50) or fails.</font>

## How the Program Works
- <font size=3>Define Functions:</font>
    - <font size=2>calculate_weighted_mark() to compute the final score.</font>
    - <font size=2>check_pass() to determine pass/fail status.</font>
- <font size=3>Read Data from File:</font>
    - <font size=2>Use the csv module to read student records.</font>
- <font size=3>Process Each Record:</font>
    - <font size=2>For each student, calculate the final mark, round it, and check if they passed.</font>
- <font size=3>Use if __name__ == "__main__":</font>
    - <font size=2>This ensures the main logic runs only when the script is executed directly.</font>

```python
#import the necessary libraries
import csv

# Function to calculate the average marks
def calc_avg_marks(ae1_marks, ae2_marks, ae1_weight=0.40, ae2_weight=0.60):
    total = (ae1_marks*ae1_weight) + (ae2_marks*ae2_weight)
    return total

# Function to check if the student passed or not
def check_pass (mark, pass_marks=50):
    if mark >= pass_marks:
        result = "Pass"
    else:
        result = "Fail"
    return result

# Function to get data from the CSV file
def GetData(filename):
    st_data = []
    try:
        with open (filename) as file:
            reader = csv.reader(file)
            next (reader) # skip the header row
            for row in reader:
                st_data.append(row)
    except Exception as e:
        print(f"An error occurred: {e}")
    
    return st_data

if __name__ == "__main__":
    # Start by getting the data from the user
    file_name = "students.csv"
    student_data = GetData(file_name)

    # For each student we need to calculate the avg marks and
    # See if the student has passed the module or not
    # Process for each student
    for student in student_data:
        student_id = student[0]
        ae1 = float(student[1])
        ae2 = float(student[2])
        final_marks = calc_avg_marks(ae1,ae2)
        rounded_marks = round(final_marks)
        result = check_pass(rounded_marks, 40)
        print (f"Student ID: {student_id}, Final marks: {rounded_marks}, Result: {result}")

```

# Modules

- <font size ="3">A module is a Python file with a .py extension that contains functions and can be imported into another Python program</font>
- <font size ="3">Python's default modules are stored in a certain location, but other modules can be stored in any location</font>
- <font size ="3">Modules can be shared and utilised to allow developers to create solutions efficiently</font>
- <font size ="3">You can bring module into a program as if you wrote them yourself</font>
- <font size ="3">You use a keyword to tell Python which modules/parts of modules you want to include in your program</font>
- <font size ="3">Modules are organised by purpose</font>

## Importing Modules

- <font size ="3">We can bring in the module into the program using:</font>
    - <font size ="3"><span style="color:green;">import</span></font>
    - <font size ="3"><span style="color:green;">from</span></font>

### import

- <font size ="3">Importing creating access to code from an external module</font>
- <font size ="3">General syntax for importing</font>
    - <font size ="3"><span style="color:green;"><b>import</b></span> [module_name] </font>
- <font size ="3">It is often convenient to rename modules with abbreviated names in code using <span style="color:blue;">as</span> after module_name. This is particularly useful where we have long module names, or we are referring to sub-modules</font>

### from

- <font size="3">Using <span style="color: green;">from</span> with <span style="color: green;">import</span> imports only the **specified** parts of a module and allows you to call imported functions without including the module name in the reference</font>

    - <font size="2"><span style="color: green;">from</span> module_name <span style="color: green;">import</span> function_name</font>

    - <font size="2"><span style="color: green;">from</span> module_name <span style="color: green;">import</span> function_name1, function_name2</font>
