## Line Charts

- <font size=3>The line graph is kind of the “hello world” of matplotlib. The following code shows how to start with a very simple line graph using the x and y-axis.</font>
- <font size=3>The function <span style="color:blue;"><b>plot()</b></span> is use for creating a line graph that connects a series of points with straight line.</font>
- <font size="3"><b>Syntax:</b></font><br>
&emsp;&emsp;&emsp;&emsp;<font size="3">plt.plot(x,y)</font>
- <font size=3>It takes two parameters; <b>a list of values representing points on the x-axis (a horizontal X axis) and a list of values representing points on the y-axis (a vertical Y axis).</b> Each point in the graph is located at a (X,Y) coordinate</font>
- <font size=3>Line chart is best when you want to show <span style="color:blue;"><b>how the value of something changes over time (trend), or compare how several things change over time relative to each other.</b></span> </font>

```python
# Importing necessary libraries
%matplotlib inline 
import matplotlib.pyplot as plt

# define the x-axis values
x = [0,2,4,6,8,10,12]

# define the y-axis values
y = [0,20,40,60,80,100,120]

# the next step is to create the figure
fig = plt.figure()

# the next step is to create the plot inside the fig container
plt.plot(x,y)

# in order to show the plot we will use plt.show() function
plt.show()
```

```python
# define the x and y-axis values
x = [0,2,4,6,8,10,12]
y = [0,20,40,60,80,100,120]

# the next step is to create the figure - Customize the figure function
fig = plt.figure(figsize=(6,4))

# the next step is to create the plot inside the fig container
plt.plot(x,y,linestyle='--', marker = '.')

# in order to show the plot we will use plt.show() function
plt.show()
```

## Bar Charts

- <font size=3>Bar chart is a rectangular display either horizontal or vertical axis.</font>
- <font size=3>The height(length) of the bar is directly proportional to the values they represent.</font>
- <font size=3>Bar charts should be used when you are showing segments of information.</font>
    - <font size=3>Vertical bar charts are useful to compare different categorical or discrete variables, such as age groups, classes, schools, etc., as long as there are not too many categories to compare. </font>
- <font size=3>Bar charts are used for <span style="color:blue;"><b>a comparison in different categories of data</b></span></font>
- <font size=3>You will use the <span style="color:blue;"><b>bar()</b></span> function for plotting the bar chart.</font>
- <font size=3>The function needs two lists: one with the X coordinates of each bar’s left edge, and another with the heights of each bar, along the Y axis. </font>
- <font size="3"><b>Syntax:</b></font><br>
&emsp;&emsp;&emsp;&emsp;<font size="3">plt.bar(x,y)</font>

```python
# For bar chats we need a list of titles and a list of values against those titles

# Create a list of titles
titles = ["Group A", "Group B", "Group C", "Group D"]

# Create a list for the values against those titles
students = [25,12,36,29]

# the next step is to create the figure - Customize the figure function
fig = plt.figure(figsize=(6,4))

# the next step is to create the bar plot inside the fig container
plt.bar(titles,students)

# setup the title of the chart
plt.title("The Students Group Data")

# setup the x-axis label
plt.xlabel("Groups Names")

# setup the y-axis label
plt.ylabel("Number of Students in Group")

# in order to show the plot we will use plt.show() function
plt.show()
```

<div class="alert alert-block alert-info">

    <li>Syntax: plt.bar(x,y, width=0.8)</li>
</ul>

```python
# Create a list of titles
titles = ["Group A", "Group B", "Group C", "Group D"]

# Create a list for the values against those titles
students = [25,12,36,29]

# the next step is to create the figure - Customize the figure function
fig = plt.figure(figsize=(5,4))

# the next step is to create the bar plot inside the fig container
plt.bar(titles,students, width=0.5)

# setup the title of the chart
plt.title("The Students Group Data")

# setup the x-axis label
plt.xlabel("Groups Names")

# setup the y-axis label
plt.ylabel("Number of Students in Group")

# in order to show the plot we will use plt.show() function
plt.show()
```

```python
# Create a list of titles
titles = ["Group A", "Group B", "Group C", "Group D"]

# Create a list for the values against those titles
students = [25,12,36,29]

# the next step is to create the figure - Customize the figure function
fig = plt.figure(figsize=(5,4))

# the next step is to create the bar plot inside the fig container - adding color and widths arg
plt.bar(titles,students, width=0.5, color=['b','r','g','y'])

# setup the title of the chart
plt.title("The Students Group Data")

# setup the x-axis label
plt.xlabel("Groups Names")

# setup the y-axis label
plt.ylabel("Number of Students in Group")

# in order to show the plot we will use plt.show() function
plt.show()
```

- <font size=3>We can create charts with multiple bars, which are typically used to compare different entities.</font>
- <font size=3>Apporach</font>
    - <font size=3>The data for plotting multiple bar charts are taken into the list.</font>
    - <font size=3>We are creating the X-axis values depending on the number of groups.</font>
    - <font size=3>To avoid overlapping of bars in each group, the bars are shifted -0.2 units and +0.2 units from the X-axis.</font>
    - <font size=3>The width of the bars of each group is taken as 0.4 units (or adjust for suitable).</font>

```python
import numpy as np

# Create a list of titles
titles = ["Group A", "Group B", "Group C", "Group D"]

# Create a list for the values against those titles
male = [20,4,18,10]
female = [5,8,18,19]

# the next step is to create the figure - Customize the figure function
fig = plt.figure(figsize=(6,4))

# create the x-axis values
x_axis = np.arange(len(titles))

# the next step is to create the bar plot inside the fig container for male students
plt.bar(x_axis-0.2,male, width=0.4, label = 'Males')

# the next step is to create the bar plot inside the fig container for female students
plt.bar(x_axis+0.2,female, width=0.4, label = 'Females')

# Labels of the x-axis
plt.xticks(x_axis,titles)

# setup the title of the chart
plt.title("The Students Group Data")

# setup the x-axis label
plt.xlabel("Groups Names")

# setup the y-axis label
plt.ylabel("Number of Students in Group")

# in order to show the plot we will use plt.show() function
plt.show()
```

- <font size=3>We can add the data label on each bar with its value</font>
- <font size=3>Apporach</font>
    - <font size=3>using plt.text() to place custom text annotations on a matplotlib plot</font>
    - <font size=3>syntax: plt.text(x,y, text, fontsize=...., ha=..., va=...) where</font>
        - <font size=3>x: Horizontal position of the text.</font>
        - <font size=3>y: Vertical position of the text.</font>
        - <font size=3>text: The string to display (e.g., the value of the bar).</font>
        - <font size=3>fontsize: Size of the text.</font>
        - <font size=3>ha / va: Horizontal and vertical alignment (e.g., 'center', 'left', 'top').</font>

```python
# Create a list of titles
titles = ["Group A", "Group B", "Group C", "Group D"]

# Create a list for the values against those titles
male = [20,4,18,10]
female = [5,8,18,19]

# the next step is to create the figure - Customize the figure function
fig = plt.figure(figsize=(6,4))

# create the x-axis values
x_axis = np.arange(len(titles))

# the next step is to create the bar plot inside the fig container for male students
line1 = plt.bar(x_axis-0.2,male, width=0.4, label = 'Males')

# the next step is to create the bar plot inside the fig container for female students
line2 = plt.bar(x_axis+0.2,female, width=0.4, label = 'Females')

# Set the y-axis beginning and ending points by using x-ticks function
plt.yticks(ticks=range(0,30,5))

# Labels of the x-axis
plt.xticks(x_axis,titles)

# Add data lables to each bar
for i in range(len(titles)):
    plt.text(x_axis[i]-0.20, male[i]-0.20, f"{male[i]}", fontsize=8,ha='center')
    plt.text(x_axis[i]+0.20, female[i]+0.20, f"{female[i]}", fontsize=8,ha='center')


# setup the title of the chart
plt.title("The Students Group Data")

# setup the x-axis label
plt.xlabel("Groups Names")

# setup the y-axis label
plt.ylabel("Number of Students in Group")

# Adding legends
#plt.legend(['Male','Female'])
plt.legend(handles=[line1,line2], loc='upper right')

# in order to show the plot we will use plt.show() function
plt.show()
```

# Scatter Plots

- <font size = 3>Scatter Plots can be used to plot data points on a horizontal and a vertical axis to display how much one variable is affected by another.</font>
- <font size = 3>Each row in the data table is represented by a dot whose position depends on its values in the columns set on the X and Y axes.</font>
- <font size = 3>Use a scatter plot to <span style="color:blue;"><b> determine whether or not two numeric features have a relationship/correlation or find a pattern. </b></span></font>
- <font size = 3>In matplotlib the <span style="color:blue;"><b>scatter()</b></span> function is used to render a scatter plot.</font>
- <font size="3"><b>Syntax:</b></font><br>
&emsp;&emsp;&emsp;&emsp;<font size="3">plt.scatter(x,y)</font>

```python
# define the x and y-axis values
x = [0,2,4,6,8,10,12]
y = [5,16,32,46,42,40,25]

# the next step is to create the figure - Customize the figure function
fig = plt.figure(figsize=(6,4))

# the next step is to create the plot inside the fig container
plt.scatter(x,y)#,linestyle='--', marker = '.')

# setup the title of the chart
plt.title("A Scatter Plot Example")

# setup the x-axis label
plt.xlabel("This is the X-Axis")

# setup the y-axis label
plt.ylabel("This is the Y-Axis")


# in order to show the plot we will use plt.show() function
plt.show()
```

- <font size = 3> You can use the following basic syntax to add a trendline to a plot in Matplotlib</font>

```python
# define the x and y-axis values
x = [0,2,4,6,8,10,12]
y = [5,16,32,46,42,40,25]

# the next step is to create the figure - Customize the figure function
fig = plt.figure(figsize=(6,4))

# the next step is to create the plot inside the fig container
plt.scatter(x,y)#,linestyle='--', marker = '.')

#Plot a simple Line Chart from the data as well
plt.plot(x,y,linestyle='--')

# add a polynomial line
z = np.polyfit(x,y,1)
p = np.poly1d(z)
plt.plot(x,p(x),linestyle='-')


# setup the title of the chart
plt.title("A Scatter Plot Example")

# setup the x-axis label
plt.xlabel("This is the X-Axis")

# setup the y-axis label
plt.ylabel("This is the Y-Axis")

# in order to show the plot we will use plt.show() function
plt.show()
```

- <font size=3>The style of the plot points (marks) can be customised. </font>

# Pie Chart

- <font size=3>The pie chart is the circular representation of the values.</font>
- <font size=3>It is divided into sectors and is directly proportional to the values.</font>
- <font size=3>Pie charts are used for displaying <span style="color:blue;"><b>a contribution of individual values to a total value</b></span></font>
- <font size=3>Pie charts are great when there are a relatively limited number of data points to examine. If you have too many, the pie chart gets sliced so many times that the visualization gives no real benefit. </font>
- <font size="3"><b>Syntax:</b></font><br>
&emsp;&emsp;&emsp;&emsp;<font size="3">plt.pie(x,labels=None)</font>
- <font size=3>When you call the pie function, you pass a list of values as an argument.</font>
    - <font size=3>The sum of the values will be used as the value of the whole.</font>
    - <font size=3>Each element in the list will become a slice in the pie chart.</font>
    - <font size=3>The size of a slice represents that element's value as a percentage of the whole.</font>
- <font size="3">The labels parameter use to display labels for the slices in the pie chart. The argument that you pass into this parameter is a list containing the desired labels, as strings.</font>

```python
# create a relevant dataset
lbl = ["Python", "Jave", "Matlab"]
percentages = [77,10,13]

# the next step is to create the figure - Customize the figure function
fig = plt.figure(figsize=(5,5))

# the next step is to create the plot inside the fig container
plt.pie(percentages,labels=lbl)

# setup the title of the chart
plt.title("A Pie Chart Example")

# setup the legends
plt.legend(loc='upper right')

# in order to show the plot we will use plt.show() function
plt.show()
```

<div class="alert alert-block alert-info">
<ul>
    <li>We can display percentage on Pie chart using keyword argument autopct</li>
        <ul><li>Syntax: plt.pie(x,labels=None, autopct='format')</li></ul>
    <li>The legend can be placed in different locations:</li>
        <ul>
            <li>"upper left", "upper right", "lower left", "lower right": Places the legend at the corresponding corner of the figure.</li>
            <li>"upper center", "lower center", "center left", "center right": Places the legend at the center of the corresponding edge of the figure.</li>
            <li>"center": Places the legend at the center of the figure.</li>
            <li>"best": Places the legend at the location with the least overlap among the defined locations so far.</li>
        </ul>
    <li>bbox_to_anchor is used to position the legend along with the loc parameter. If you provide a 2-tuple (x, y), it places the corner of the legend specified by loc at the x, y coordinates.</li>
</ul>
</div>

```python
# create a relevant dataset
lbl = ["Python", "Jave", "Matlab"]
percentages = [77.33,10.66,12.01]

# the next step is to create the figure - Customize the figure function
fig = plt.figure(figsize=(5,5))

# the next step is to create the plot inside the fig container
plt.pie(percentages,labels=lbl, autopct="%1.1f%%")

# setup the title of the chart
plt.title("A Pie Chart Example")

# setup the legends
plt.legend(loc='upper right')

# in order to show the plot we will use plt.show() function
plt.show()
```

## Yearly trend in the average budget for each movie genre from 2010-2016

```python
# select target columns
selected_df = df[['year', 'budgetmillions','genres']]
selected_df
```

```python
# next slice will be in the row level from 2010 to 2016
year_2010_2016_df = selected_df.loc[(selected_df['year'] >= 2010) & (selected_df['year'] <= 2016)]
year_2010_2016_df
```

```python
# calculate average budget of movies from 2010 to 2016
year_avg_budget = year_2010_2016_df.groupby(['year','genres'])['budgetmillions'].mean().reset_index()
year_avg_budget
```

```python
# Create figure object and set up the size of figure
fig = plt.figure(figsize=(15,8)) 

# Plot a line chart showing the average budget (in millions) for each movie genre.
for genre in year_avg_budget['genres'].unique():
    data = year_avg_budget.loc[year_avg_budget['genres'] == genre]
    plt.plot(data['year'],data['budgetmillions'], label=genre)

#set label and title
plt.xlabel('Year')
plt.ylabel('Money (millions)')
plt.title('Average yearly budget spent on movies by genre.')
#plt.legend() # add the legend
plt.show() # show the graph
```
